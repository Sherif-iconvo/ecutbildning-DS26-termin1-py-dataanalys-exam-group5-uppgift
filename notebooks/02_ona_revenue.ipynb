```python
df = pd.read_csv('../data/ecommerce_sales.csv', encoding='utf-8')
df['date'] = pd.to_datetime(df['date'])
```

```python
import pandas as pd
import sys

# Lägg till src-mappen i sökvägen
sys.path.append("../src")

from metrics import revenue_by_city

# Beräkna total intäkt per stad
city_revenue = revenue_by_city(df)

# Visa resultatet
print("Totala intäkter per stad:")
display(city_revenue)
```

# Diagram – Total intäkt per stad

Diagrammet visar vilka städer som genererar högst intäkt under den analyserade perioden.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

city_revenue.plot(
    kind="bar",
    color="skyblue",
    edgecolor="navy"
)

plt.title("Total intäkt per stad")
plt.xlabel("Stad")
plt.ylabel("Intäkt (SEK)")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```
