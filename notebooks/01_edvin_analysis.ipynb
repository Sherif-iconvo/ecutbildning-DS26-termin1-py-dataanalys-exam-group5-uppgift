# Loading data into a dataframe.

```python
import pandas as pd

df = pd.read_csv("../data/ecommerce_sales.csv")

print(df)
```

First control off the data

```python
#Gives us info about the columns:
print(df.info(), "\n")
#Gives us number of non-values in each column:                      
print(df.isna().sum(), "\n")
#Checking wheter we have reasonable data.
print(df.describe(), "\n")
#Check wheter there is duplicates in any of the rows.                
print("Dubletter:", df.duplicated().sum())  
```

Using a seperated dataframe to not adjust in the original file

```python
df_clean = df.copy()
```

Doing necessary changes to clean up the data.

```python
df_clean["date"] = pd.to_datetime(df_clean["date"], dayfirst=False, errors="coerce")
```

Doing a last check that our data is totally cleaned.

```python
print(df_clean.info(), "\n")                   
print(df_clean.isna().sum(), "\n")                
print("Dubletter:", df_clean.duplicated().sum())  
```
