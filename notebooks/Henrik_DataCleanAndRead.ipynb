```python
import numpy as np
import pandas as pd


df = pd.read_csv("../data/ecommerce_sales.csv")


#General information overview about the data in the file
print(df.info(), "\n")

#Shows all the columns/index
print(df.columns, "\n")

#Looking for missing values in the data
print(df.isna().sum() , "\n")

#Looking for duplicated values in the data
print("duplicates:",df.duplicated().sum(), "\n")

#Overview of the data
df.head()

#Creates a copy of the dataframe
df_clean = df.copy()

#Converts the "date" column to a date/time-objekt
df_clean["date"] = pd.to_datetime(df_clean["date"], errors="coerce")

print()

#Check to see if the conversion was successful
print(df_clean.dtypes)
```
