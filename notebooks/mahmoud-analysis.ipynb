## Kontroll av datakvalitet

```python
import pandas as pd 
```

```python
table = pd.read_csv("../data/ecommerce_sales.csv")
table_clean = table.copy() 
```

```python
table_clean.info()
```

```python
table_clean["date"] = pd.to_datetime(table_clean["date"], errors="coerce")

table_clean["city"] = table_clean["city"].astype("category")
table_clean["category"] = table_clean["category"].astype("category")
```

```python
table_clean.dtypes
```

```python
table_clean["date"].isna().sum() 
```

```python
print("First date", table_clean["date"].min()) 
print("Last date:", table_clean["date"].max())
```

```python
(table_clean["price"] <= 0).sum() 
```

```python
table_clean.isna().sum() 
```

```python
table_clean["city"].value_counts(dropna=False) 
```

```python
table_clean["category"].value_counts(dropna=False) 
```

```python
table_clean["units"].value_counts(dropna=False) 
```

```python
table_clean.duplicated().sum() 
```

```python
table_clean["order_id"].duplicated().sum() 
```

```python
expected_revenue = table_clean["price"] * table_clean["units"] 
```

```python
difference = (table_clean["revenue"] - expected_revenue)
```
