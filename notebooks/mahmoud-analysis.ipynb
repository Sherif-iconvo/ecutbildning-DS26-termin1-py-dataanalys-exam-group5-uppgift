## Kontroll av datakvalitet

```python
import pandas as pd # för att kunna läsa in csv-filer och arbeta med dataframes
```

```python
table = pd.read_csv("../data/ecommerce_sales.csv") # för att läsa in csv-filen och spara den som en dataframe
```

```python
table[table["price"] < 0] # för att kolla vilka rader som har ett pris som är mindre än 0
```

```python
(table["price"] <= 0).sum() # för att kolla hur många rader som har ett pris som är mindre än eller lika med 0
```

```python
table.isna().sum() # för att kolla om det finns några NaN-värden i datasetet
```

```python
table["city"].value_counts(dropna=False) # för att kolla hur många unika värden som finns i kolumnen "city" och hur många gånger varje värde förekommer, inklusive NaN-värden
```

```python
table["category"].value_counts(dropna=False) # för att kolla hur många unika värden som finns i kolumnen "category" och hur många gånger varje värde förekommer, inklusive NaN-värden
```

```python
table["units"].value_counts(dropna=False) # för att kolla hur många unika värden som finns i kolumnen "units" och hur många gånger varje värde förekommer, inklusive NaN-värden
```

```python
table.duplicated().sum() # för att kolla om det finns några dubbletter i datasetet
```

```python
table["order_id"].duplicated().sum() # för att kolla om det finns några dubbletter i kolumnen "order_id"
```

```python
dates = pd.to_datetime(table["date"], errors="coerce") # för att konvertera kolumnen "date" till datetime-format och hantera eventuella felaktiga värden
```

```python
dates.isna().sum() # för att kolla hur många rader som har ett datum som inte kunde konverteras till datetime-format (NaT-värden)
```

```python
print("أول تاريخ:", dates.min()) # för att skriva ut det tidigaste datumet i kolumnen "date"
print("آخر تاريخ:", dates.max()) # för att skriva ut det senaste datumet i kolumnen "date"
```

```python
expected_revenue = table["price"] * table["units"]
```

```python
difference = (table["revenue"] - expected_revenue).abs()
```

```python
table[difference > 0.01]
```
