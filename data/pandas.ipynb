```python
import sys; sys.path.append("..")
import pandas as pd
import matplotlib.pyplot as plt
from src.ecommerce import EcommerceAnalyzer

# df = pd.read_csv("ecommerce_sales.csv", parse_dates=["date"])
```

```python
df = EcommerceAnalyzer().df

print(df["date"].min(), "->",df["date"].max())              # period covered
print(type(df["date"].min()))
print()

df_temp = df.copy()
df_temp["period"] = df_temp["date"].dt.to_period("M")
monthly_orders = df_temp["period"].value_counts().sort_index()      # orders per month
monthly_orders.name = "month_orders"
print(monthly_orders)
print()

weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday",]
weekday_orders = df["date"].dt.day_name().value_counts().reindex(weekdays) # orders per weekday
weekday_orders.name = "Weekday_orders"
print(weekday_orders) 
print(type(weekday_orders))
print()

df_temp = df.copy()
df_temp["period"] = df_temp["date"].dt.to_period("M")
monthly = df_temp.groupby(["period"])["revenue"].agg(["sum", "mean", "count"])    # sum: total revenue per month/ mean: average order value (AOV) per month / count: orders per month
print(monthly)
print(type(monthly))
print()
```

```python
category_performance = (df.groupby(["category"])                    # total revenue, units sold, and order count per product category
                   .agg
                   (
                       total_revenue = ("revenue", "sum"), 
                       units_sold = ("units", "sum"), 
                       order_count = ("order_id", "count"),
                       )
                       .sort_values(by="total_revenue", ascending=False)
                )

print(category_performance)

print()
df["period"] = df["date"].dt.to_period("M")
category_performance = (df.groupby(["period", "category"])                    # total revenue, units sold, and order count per product category per month
                   .agg
                   (
                       total_revenue = ("revenue", "sum"), 
                       units_sold = ("units", "sum"), 
                       order_count = ("order_id", "count"),
                       )
                       .sort_values(by=["period", "total_revenue"], ascending=[True, False])
                )

print(category_performance)
```

```python
city_performance = (df.groupby(["city"])                    # total revenue, units sold, and average order value per product city
                   .agg
                   (
                       total_revenue = ("revenue", "sum"), 
                       units_sold = ("units", "sum"), 
                       aov = ("revenue", "mean"),
                       )
                       .sort_values(by="total_revenue", ascending=False)
                )

print(city_performance)


print()
df["period"] = df["date"].dt.to_period("M")
city_performance = (df.groupby(["period", "city"])           # total revenue, units sold, and average order value per product city per month
                   .agg
                   (
                       total_revenue = ("revenue", "sum"), 
                       units_sold = ("units", "sum"), 
                       aov = ("revenue", "mean"),
                       )
                       .sort_values(by=["period", "total_revenue"], ascending=[True, False])
                )

print(city_performance)
```
